```python
import requests

api_url = "https://www.geoboundaries.org/api/current/gbOpen/MAR/ADM1/"

try:
    meta = requests.get(api_url, timeout=60).json()
    for key in ["boundaryName", "boundaryYearRepresented", "boundarySource",
                "boundaryLicense", "simplifiedGeometryGeoJSON"]:
        print(f"{key} : {meta.get(key)}")

    geo = requests.get(meta["simplifiedGeometryGeoJSON"], timeout=120)
    print("Statut du fichier :", geo.status_code, "| Taille :", len(geo.content), "octets")

    features = geo.json()["features"]
    print("Nombre de régions :", len(features))
    print(sorted(f["properties"].get("shapeName") for f in features))
except Exception as error:
    print("Erreur :", type(error).__name__, error)
```
