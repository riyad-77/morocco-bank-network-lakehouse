```python
from pyspark.sql import functions as F

df = spark.table("workspace.bank_lakehouse.bronze_osm_banks") \
    .filter(F.col("_ingest_date") == "2026-11-06")

print("Valeurs de brand :")
df.groupBy(F.col("tags")["brand"].alias("brand")) \
  .count().orderBy(F.desc("count")).show(40, truncate=False)

print("Noms les plus fréquents quand brand est vide :")
df.filter(F.col("tags")["brand"].isNull()) \
  .groupBy(F.col("tags")["name"].alias("name")) \
  .count().orderBy(F.desc("count")).show(40, truncate=False)
```

```python
display(
    df.filter(F.col("tags")["brand"].isNull())
      .groupBy(F.col("tags")["name"].alias("name"))
      .count()
      .orderBy(F.desc("count"))
      .limit(50)
)
```
